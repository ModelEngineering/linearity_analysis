```python
import numpy as np
import pandas as pd
import tellurium as te
```

# Oscillator

```python
MODEL = """
$S8 -> S6; k0*S8
S1 -> S3; k2*S1
S3 + S6 ->; k4*S3*S6
S4 -> S1; k5*S4
S6 -> S6 + S4; k6*S6

k = 0.1
k0 = k*24; k2 = k*15; k3 = k*54; k4 = k*315
k5 = k*24; k6 = k*53
S8 = k*6.0
"""
rr = te.loada(MODEL)
rr.simulate(0, 500, 1000)
rr.plot()
```
