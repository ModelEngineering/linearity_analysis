```python
import pandas as pd
import numpy as np
import tellurium as te
```

# Sequential Network

```python
model  = """
 -> S1; k0
S1 -> S2; k1*S1
S2 -> S3; k2*S2
S3 -> ; k3*S2


S0 = 10
S1 = 0
S2 = 0
S3 = 0
k0=1; k1 = 1; k2 = 1; k3 = 0.8
"""
rr = te.loada(model)
rr.simulate(0, 20, 200)
rr.plot()
```

# Stability Analysis

```python
MODEL  = """
 -> S1; k0
S1 ->  S2; k1*S1
S2 -> S1; k2*S2
S2 -> S3; k3*S2
S3 -> S2; k4*S3
S3 -> ; k5*S3

// Stability remediations
S2 -> ; 0.3*S2
S3 -> ; 0.1*S3


S0 = 10
S1 = 0
S2 = 0
S3 = 0
k0=1; k1 = 1; k2 = 0.8; k3 = 0.5; k4 = 0.5; k5 = 0
"""
rr = te.loada(MODEL)
rr.simulate(0, 50, 500)
rr.plot()
#rr.getFullJacobian(), rr.getFullEigenValues()
```
